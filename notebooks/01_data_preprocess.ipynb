# 01 数据预处理
加载沪深300数据，清洗，计算因子和未来收益

```python
import sys, os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.insert(0, os.path.abspath('..'))
from src.data_loader import load_data, calc_forward_return
from src.factors import calc_factors, process_factors

%matplotlib inline
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
```

```python
df = load_data()
df = calc_forward_return(df, periods=5)
df = calc_factors(df)
df = process_factors(df)

print(f'数据维度: {df.shape}')
print(f'时间范围: {df["date"].min()} ~ {df["date"].max()}')
print(f'股票数量: {df["code"].nunique()}')
df.head()
```

```python
# 缺失值检查
na_counts = df.isna().sum()
na_counts[na_counts > 0]
```

```python
# 因子分布可视化
factor_cols = ['factor_pe', 'factor_momentum_20', 'factor_vol_60']
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, factor_cols):
    # 随机抽样10000个点避免过载
    sample = df[col].dropna().sample(min(10000, len(df)), random_state=42)
    ax.hist(sample, bins=80, alpha=0.7)
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

```python
# 保存处理后的数据
df.to_parquet('../data/processed_data.parquet', index=False)
print('数据保存完成')
```
